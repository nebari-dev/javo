# Times table

Move the slider: the sentence and the table update.

```python
import ipywidgets as widgets
from IPython.display import display

number = widgets.IntSlider(value=7, min=1, max=20, description="Number")
sentence = widgets.HTML()
table = widgets.HTML()


def update(change=None):
    n = number.value
    sentence.value = f"<p>{n} squared is <b>{n * n}</b>.</p>"
    rows = "".join(f"<tr><td>{i} x {n}</td><td>{i * n}</td></tr>" for i in range(1, 11))
    table.value = f"<table>{rows}</table>"


number.observe(update, names="value")
update()
display(number, sentence, table)
```
