## 2. Draw a chart

The cell below builds a small bar chart as an SVG image with the standard library. Change the numbers and run it again.


```python
from IPython.display import SVG

sales = {"North": 120, "South": 95, "East": 150, "West": 70}
width, height, bar, gap = 360, 200, 60, 20
top = max(sales.values())
parts = [f'<svg xmlns="http://www.w3.org/2000/svg" width="{width}" height="{height}">']
for i, (region, value) in enumerate(sales.items()):
    h = int(140 * value / top)
    x = 20 + i * (bar + gap)
    parts.append(f'<rect x="{x}" y="{160 - h}" width="{bar}" height="{h}" fill="#4a5568"/>')
    parts.append(f'<text x="{x + bar / 2}" y="180" text-anchor="middle" font-size="12">{region}</text>')
    parts.append(f'<text x="{x + bar / 2}" y="{154 - h}" text-anchor="middle" font-size="12">{value}</text>')
parts.append("</svg>")
SVG("".join(parts))
```
